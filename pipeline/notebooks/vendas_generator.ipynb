```python
import pandas as pd
import random
from faker import Faker
from pathlib import Path

# Função para encontrar o diretório raiz do repositório
def find_repo_root(start: Path = Path.cwd()) -> Path:
    p = start.resolve()
    for parent in [p] + list(p.parents):
        if (parent / "README.md").exists() or (parent / ".git").exists():
            return parent
    return p  # fallback: cwd


BASE_DIR = find_repo_root()
RAW_DIR = BASE_DIR / "data" / "raw"
RAW_DIR.mkdir(parents=True, exist_ok=True)
vendas_raw = RAW_DIR / "vendas_raw.csv"
```

```python
# Gera o objeto Faker com localidade brasileira
fake = Faker("pt_BR")


# Gera uma lista dicionario e popula ela com dados falsos da bibli. faker para gerar uma tabela de dados 
vendas = []

for i in range(5000):
    vendas.append({
        # Identificadores
        "id_venda": i + 1,
        "id_pedido": f"PED-{random.randint(10000, 99999)}",
        
        # Datas (formatos variados de propósito)
        "data_venda": fake.date_between(start_date="-2y", end_date="today"),
        "data_envio": fake.date_between(start_date="-2y", end_date="today"),
        
        # Cliente
        "id_cliente": random.randint(1, 1200),
        "nome_cliente": fake.name(),
        "email_cliente": fake.email(),
        "cidade_cliente": fake.city(),
        "estado_cliente": fake.state_abbr(),
        
        # Produto
        "categoria_produto": random.choice(
            ["Eletrônicos", "eletronicos", "Roupas", "Casa", "Beleza"]
        ),
        "produto": fake.word().capitalize(),
        "marca": random.choice(
            ["Samsung", "Apple", "LG", "Nike", "Adidas", "Genérica"]
        ),
        
        # Valores (alguns campos redundantes)
        "quantidade": random.randint(1, 5),
        "valor_unitario": round(random.uniform(50, 3500), 2),
        "valor_venda": round(random.uniform(50, 3500), 2),
        "desconto": random.choice([0, 0, 0, 10, 20, 30]),
        "frete": round(random.uniform(0, 120), 2),
        
        # Canal (não padronizado de propósito)
        "canal": random.choice(
            ["Site", "SITE", "App", "Aplicativo", "app"]
        ),
        
        # Pagamento
        "forma_pagamento": random.choice(
            ["Crédito", "Débito", "Pix", "Boleto"]
        ),
        "parcelas": random.choice([1, 1, 1, 2, 3, 6, 12]),
        "status_pagamento": random.choice(
            ["Aprovado", "aprovado", "Recusado", "Pendente"]
        ),
        
        # Logística
        "transportadora": random.choice(
            ["Correios", "Jadlog", "Total Express"]
        ),
        "status_entrega": random.choice(
            ["Entregue", "Em trânsito", "Atrasado"]
        )
    })
```

```python
df_raw = pd.DataFrame(vendas)

#df_raw
```

```python
df_raw.to_csv(
    vendas_raw,
    index=False,
    encoding="utf-8"
)
```
