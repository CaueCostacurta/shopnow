```python
import sys
import pandas as pd
from pathlib import Path

# adiciona 'pipeline/' ao path
sys.path.append(str(Path.cwd().parent.parent))
from pipeline.utils import get_raw_dir

RAW_DIR = get_raw_dir()
```

```python

def carregar_dados_csv():
    """Carrega os CSVs raw e retorna (metas, vendas) como DataFrames."""

    # Define os caminhos dos arquivos CSV
    metas_path = RAW_DIR / "metas_vendas_raw.csv"
    vendas_path = RAW_DIR / "vendas_raw.csv"

    try:
        # Carrega os dados dos arquivos CSV em DataFrames do pandas
        df_metas = pd.read_csv(metas_path)
        df_vendas = pd.read_csv(vendas_path)
        return df_metas, df_vendas
    
    except FileNotFoundError as e:
        if not metas_path.exists():
            raise FileNotFoundError(f"O arquivo {metas_path} não foi encontrado.") from e
        if not vendas_path.exists():
            raise FileNotFoundError(f"O arquivo {metas_path} não foi encontrado.") from e
        raise 

```

```python
df_metas, df_vendas = carregar_dados_csv()
#df_metas
df_vendas
```
