```python
import sys
from pathlib import Path
import random
import pandas as pd
from faker import Faker
from datetime import datetime, timedelta

# Caminho real do notebook 
NOTEBOOK_DIR = Path.cwd() 

# Sobe até a raiz do projeto (shopnow/) 
ROOT = NOTEBOOK_DIR.parents[2] 

# Adiciona ao sys.path 
if str(ROOT) not in sys.path: sys.path.append(str(ROOT)) 

# Agora sim, pode importar 
from pipeline.utils import find_repo_root, get_raw_dir

# 1. Configuração do ambiente

ROOT = find_repo_root()
RAWDIR = get_raw_dir()
```

```python
# 2. Configuração do Faker, Seed e geração dos dados

# Inicializa o Faker e a seed para reprodutibilidade
fake = Faker("pt_BR")
random.seed(42)

def gerar_rh(qtd: int = 350) -> list[dict]:
    deps = ["TI", "ti", "Marketing", "Operações", "Financeiro"]

    rh = []

    for i in range(qtd):
        adm = fake.date_between("-6y")

        rh.append({
            "id_funcionario": i + 1,
            "nome": fake.name(),
            "departamento": random.choice(deps),
            "cargo": fake.job(),
            "data_admissao": adm,
            "salario": random.randint(2500, 22000),
            "beneficios": random.randint(300, 2500),
            "bonus": random.choice([0, 0, 1000, 2000])
        })

    return rh
```

```python
# 3. Gera o DataFrame
df_rh = pd.DataFrame(gerar_rh(350))
#df_rh
```

```python
# 4. Salva o DataFrame como CSV no diretório de dados brutos
df_rh.to_csv(
    RAWDIR / "rh_raw.csv",
    index=False,
    encoding="utf-8"
)
```
