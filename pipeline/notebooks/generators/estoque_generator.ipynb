```python
import sys
from pathlib import Path
import random
import pandas as pd
from faker import Faker
from datetime import datetime, timedelta

# Caminho real do notebook 
NOTEBOOK_DIR = Path.cwd() 

# Sobe até a raiz do projeto (shopnow/) 
ROOT = NOTEBOOK_DIR.parents[2] 

# Adiciona ao sys.path 
if str(ROOT) not in sys.path: sys.path.append(str(ROOT)) 

# Agora sim, pode importar 
from pipeline.utils import find_repo_root, get_raw_dir

# 1. Configuração do ambiente

ROOT = find_repo_root()
RAWDIR = get_raw_dir()
```

```python
# 2. Configuração do Faker, Seed e geração dos dados

# Inicializa o Faker e a seed para reprodutibilidade
fake = Faker("pt_BR")
random.seed(42)

def gerar_estoque(qtd: int = 800) -> list[dict]:
    cds = ["CD-SP", "cd_sp", "CD-RJ", "CD-MG"]

    estoque = []

    for i in range(qtd):
        atual = random.randint(-20, 600)

        estoque.append({
            "id_estoque": i + 1,
            "id_produto": i + 1,
            "centro_distribuicao": random.choice(cds),
            "estoque_atual": atual,
            "estoque_reservado": random.randint(0, 80),
            "estoque_minimo": random.randint(5, 50),
            "data_posicao": fake.date_between("-30d")
        })

    return estoque
```

```python
# 3. Gera o DataFrame
df_estoque = pd.DataFrame(gerar_estoque())
#df_estoque
```

```python
# 4. Salva o DataFrame como CSV no diretório de dados brutos
df_estoque.to_csv(
    RAWDIR / "estoque_raw.csv",
    index=False, encoding="utf-8")
```
