```python
import sys
from pathlib import Path
import random
import pandas as pd
from faker import Faker
from datetime import datetime, timedelta

# Caminho real do notebook 
NOTEBOOK_DIR = Path.cwd() 

# Sobe até a raiz do projeto (shopnow/) 
ROOT = NOTEBOOK_DIR.parents[2] 

# Adiciona ao sys.path 
if str(ROOT) not in sys.path: sys.path.append(str(ROOT)) 

# Agora sim, pode importar 
from pipeline.utils import find_repo_root, get_raw_dir

# 1. Configuração do ambiente

ROOT = find_repo_root()
RAWDIR = get_raw_dir()
```

```python
# 2. Configuração do Faker, Seed e geração dos dados

# Inicializa o Faker e a seed para reprodutibilidade
fake = Faker("pt_BR")
random.seed(42)

def gerar_produtos(qtd: int = 800) -> list[dict]:
    categorias = ["Eletrônicos", "eletronicos", "Roupas", "Casa", "Beleza"]
    marcas = ["Samsung", "Apple", "LG", "Nike", "Adidas", "Genérica"]

    produtos = []

    for i in range(qtd):
        preco = random.uniform(30, 4000)

        produtos.append({
            "id_produto": i + 1,
            "sku": f"SKU-{random.randint(1000,9999)}",
            "nome_produto": fake.word().capitalize(),
            "categoria": random.choice(categorias),
            "marca": random.choice(marcas),
            "preco_lista": round(preco, 2),
            "custo_unitario": round(preco * random.uniform(0.3, 0.8), 2),
            "peso_kg": round(random.uniform(0.1, 15), 2),
            "ativo": random.choice([True, True, True, False])
        })

    return produtos
```

```python
# 3. Gera o DataFrame
df_produtos = pd.DataFrame(gerar_produtos())
#df_produtos
```

```python
# 4. Salva o DataFrame como CSV no diretório de dados brutos
df_produtos.to_csv(
    RAWDIR / "produtos_raw.csv",
    index=False, encoding="utf-8")
```
