```python
import sys
from pathlib import Path
import random
import pandas as pd
from faker import Faker
from datetime import datetime, timedelta

# Caminho real do notebook 
NOTEBOOK_DIR = Path.cwd() 

# Sobe até a raiz do projeto (shopnow/) 
ROOT = NOTEBOOK_DIR.parents[2] 

# Adiciona ao sys.path 
if str(ROOT) not in sys.path: sys.path.append(str(ROOT)) 

# Agora sim, pode importar 
from pipeline.utils import find_repo_root, get_raw_dir

# 1. Configuração do ambiente

ROOT = find_repo_root()
RAWDIR = get_raw_dir()
```

```python
# 2. Configuração do Faker, Seed e geração dos dados

# Inicializa o Faker e a seed para reprodutibilidade
fake = Faker("pt_BR")
random.seed(42)

def gerar_fornecedores(qtd: int = 120) -> list[dict]:
    forn = []

    for i in range(qtd):
        forn.append({
            "id_fornecedor": i + 1,
            "nome": fake.company(),
            "cidade": fake.city(),
            "estado": random.choice([fake.state_abbr(), fake.state()])
        })

    return forn
```

```python
# 3. Gera o DataFrame
df_fornecedores = pd.DataFrame(gerar_fornecedores(120))
#df_fornecedores
```

```python
# 4. Salva o DataFrame como CSV no diretório de dados brutos
df_fornecedores.to_csv(
    RAWDIR / "fornecedores_raw.csv",
    index=False,
    encoding="utf-8"
)
```
