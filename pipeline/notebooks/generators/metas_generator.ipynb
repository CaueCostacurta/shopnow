```python
import sys
import pandas as pd
import random
from pathlib import Path

try: 
    ROOT = Path(__file__).resolve().parents[2] 
except NameError: 
    # estamos em um notebook 
    ROOT = Path.cwd().parents[2] 
sys.path.append(str(ROOT))

from pipeline.utils import create_raw_dir

# Cria o diretório de dados brutos, se não existir
RAW_DIR = create_raw_dir()
```

```python
def gerar_metas_vendas() -> list[dict]:
    meses = pd.period_range("2023-01", "2024-12", freq="M")
    categorias = ["Eletrônicos", "Roupas", "Casa", "Beleza"]
    canais = ["Site", "App"]

    metas = []

    for mes in meses:
        for categoria in categorias:
            for canal in canais:
                metas.append({
                    "ano_mes": str(mes),
                    "categoria_produto": categoria,
                    "canal": canal,
                    "meta_faturamento": random.randint(200000, 800000)
                })
    return metas
```

```python
df_metas = pd.DataFrame(gerar_metas_vendas())

#df_metas
```

```python
df_metas.to_csv(
    RAW_DIR / "metas_raw.csv",
    index=False,
    encoding="utf-8"
)
```
