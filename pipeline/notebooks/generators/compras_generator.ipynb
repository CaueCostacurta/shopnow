```python
import sys
from pathlib import Path
import random
import pandas as pd
from faker import Faker
from datetime import datetime, timedelta

# Caminho real do notebook 
NOTEBOOK_DIR = Path.cwd() 

# Sobe até a raiz do projeto (shopnow/) 
ROOT = NOTEBOOK_DIR.parents[2] 

# Adiciona ao sys.path 
if str(ROOT) not in sys.path: sys.path.append(str(ROOT)) 

# Agora sim, pode importar 
from pipeline.utils import find_repo_root, get_raw_dir

# 1. Configuração do ambiente

ROOT = find_repo_root()
RAWDIR = get_raw_dir()
```

```python
# 2. Configuração do Faker, Seed e geração dos dados

# Inicializa o Faker e a seed para reprodutibilidade
fake = Faker("pt_BR")
random.seed(42)


def gerar_compras(qtd: int = 1800) -> list[dict]:
    compras = []

    for i in range(qtd):
        qtd_c = random.randint(5, 600)
        preco = random.uniform(8, 900)

        compras.append({
            "id_compra": i + 1,
            "id_fornecedor": random.randint(1, 120),
            "id_produto": random.randint(1, 800),
            "data_compra": fake.date_between("-2y"),
            "quantidade": qtd_c,
            "preco_unitario": round(preco, 2),
            "valor_total": round(qtd_c * preco, 2)
        })

    return compras
```

```python
# 3. Gera o DataFrame
df_compras = pd.DataFrame(gerar_compras(1800))
#df_compras
```

```python
# 4. Salva o DataFrame como CSV no diretório de dados brutos
df_compras.to_csv(
    RAWDIR / "compras_raw.csv",
    index=False,
    encoding="utf-8"
)
```
