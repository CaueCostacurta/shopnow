```python
import sys
from pathlib import Path
import random
import pandas as pd
from faker import Faker
from datetime import datetime, timedelta

# Caminho real do notebook 
NOTEBOOK_DIR = Path.cwd() 

# Sobe até a raiz do projeto (shopnow/) 
ROOT = NOTEBOOK_DIR.parents[2] 

# Adiciona ao sys.path 
if str(ROOT) not in sys.path: sys.path.append(str(ROOT)) 

# Agora sim, pode importar 
from pipeline.utils import find_repo_root, get_raw_dir

# 1. Configuração do ambiente

ROOT = find_repo_root()
RAWDIR = get_raw_dir()
```

```python
# 2. Configuração do Faker, Seed e geração dos dados

# Inicializa o Faker e a seed para reprodutibilidade
fake = Faker("pt_BR")
random.seed(42)

def gerar_financeiro(qtd: int = 7000) -> list[dict]:
    categorias = ["Marketing", "marketing", "Logística", "RH", "Tecnologia", "Receita"]

    lanc = []

    for i in range(qtd):
        tipo = random.choice(["Receita", "receita", "Despesa"])

        lanc.append({
            "id_lancamento": i + 1,
            "data": fake.date_between("-2y"),
            "tipo": tipo,
            "categoria": random.choice(categorias),
            "descricao": fake.sentence(nb_words=4),
            "valor": round(random.uniform(-8000, 12000), 2),
            "id_pedido": random.choice([None, f"PED-{random.randint(10000,99999)}"])
        })

    return lanc
```

```python
# 3. Gera o DataFrame
df_financeiro = pd.DataFrame(gerar_financeiro(7000))
#df_financeiro
```

```python
# 4. Salva o DataFrame como CSV no diretório de dados brutos
df_financeiro.to_csv(
    RAWDIR / "financeiro_raw.csv",
    index=False,
    encoding="utf-8"
)
```
