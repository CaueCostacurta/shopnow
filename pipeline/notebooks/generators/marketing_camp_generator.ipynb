```python
import sys
from pathlib import Path
import random
import pandas as pd
from faker import Faker
from datetime import datetime, timedelta

# Caminho real do notebook 
NOTEBOOK_DIR = Path.cwd() 

# Sobe até a raiz do projeto (shopnow/) 
ROOT = NOTEBOOK_DIR.parents[2] 

# Adiciona ao sys.path 
if str(ROOT) not in sys.path: sys.path.append(str(ROOT)) 

# Agora sim, pode importar 
from pipeline.utils import find_repo_root, get_raw_dir

# 1. Configuração do ambiente

ROOT = find_repo_root()
RAWDIR = get_raw_dir()
```

```python
# 2. Configuração do Faker, Seed e geração dos dados

# Inicializa o Faker e a seed para reprodutibilidade
fake = Faker("pt_BR")
random.seed(42)

def gerar_campanhas(qtd: int = 60) -> list[dict]:
    canais = ["Google", "google", "Meta", "Facebook", "TikTok", "Email"]

    campanhas = []

    for i in range(qtd):
        ini = fake.date_between("-1y")
        fim = ini + timedelta(days=random.randint(15, 120))

        campanhas.append({
            "id_campanha": i + 1,
            "nome_campanha": fake.bs(),
            "canal": random.choice(canais),
            "tipo": random.choice(["Aquisição", "aquisicao", "Remarketing"]),
            "data_inicio": ini,
            "data_fim": fim,
            "orcamento": random.randint(5000, 250000)
        })

    return campanhas
```

```python
# 3. Gera o DataFrame
df_marketing_camp = pd.DataFrame(gerar_campanhas())
#df_marketing_camp
```

```python
# 4. Salva o DataFrame como CSV no diretório de dados brutos

df_marketing_camp.to_csv(
    RAWDIR / "marketing_campanhas_raw.csv",
    index=False, encoding="utf-8")
```
