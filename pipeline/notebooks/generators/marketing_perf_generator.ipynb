```python
import sys
from pathlib import Path
import random
import pandas as pd
from faker import Faker
from datetime import datetime, timedelta

# Caminho real do notebook 
NOTEBOOK_DIR = Path.cwd() 

# Sobe até a raiz do projeto (shopnow/) 
ROOT = NOTEBOOK_DIR.parents[2] 

# Adiciona ao sys.path 
if str(ROOT) not in sys.path: sys.path.append(str(ROOT)) 

# Agora sim, pode importar 
from pipeline.utils import find_repo_root, get_raw_dir

# 1. Configuração do ambiente

ROOT = find_repo_root()
RAWDIR = get_raw_dir()
```

```python
# 2. Configuração do Faker, Seed e geração dos dados

# Inicializa o Faker e a seed para reprodutibilidade
fake = Faker("pt_BR")
random.seed(42)

def gerar_marketing_performance(qtd: int = 2500) -> list[dict]:
    perf = []

    for i in range(qtd):
        cliques = random.randint(50, 15000)

        perf.append({
            "id_registro": i + 1,
            "data": fake.date_between("-1y"),
            "id_campanha": random.randint(1, 60),
            "impressoes": random.randint(500, 80000),
            "cliques": cliques,
            "conversoes": random.randint(0, cliques + 20),
            "custo": round(random.uniform(50, 9000), 2)
        })

    return perf
```

```python
# 3. Gera o DataFrame
df_marketing_perf = pd.DataFrame(gerar_marketing_performance())
#df_marketing_perf
```

```python
# 4. Salva o DataFrame como CSV no diretório de dados brutos
df_marketing_perf.to_csv(
    RAWDIR / "marketing_performance_raw.csv",
    index=False, encoding="utf-8")
```
