```python
import sys
from pathlib import Path
import random
import pandas as pd
from faker import Faker
from datetime import datetime, timedelta

# Caminho real do notebook 
NOTEBOOK_DIR = Path.cwd() 

# Sobe até a raiz do projeto (shopnow/) 
ROOT = NOTEBOOK_DIR.parents[2] 

# Adiciona ao sys.path 
if str(ROOT) not in sys.path: sys.path.append(str(ROOT)) 

# Agora sim, pode importar 
from pipeline.utils import find_repo_root, get_raw_dir

# 1. Configuração do ambiente

ROOT = find_repo_root()
RAWDIR = get_raw_dir()
```

```python
# 2. Configuração do Faker, Seed e geração dos dados

# Inicializa o Faker e a seed para reprodutibilidade
fake = Faker("pt_BR")
random.seed(42)

def gerar_clientes(qtd: int = 1500) -> list[dict]:
    canais = ["Ads", "ads", "Orgânico", "Organico", "Social"]
    tipos = ["PF", "pf", "PJ"]

    clientes = []

    for i in range(qtd):
        nasc = fake.date_of_birth(minimum_age=18, maximum_age=85)

        clientes.append({
            "id_cliente": i + 1,
            "nome": fake.name(),
            "email": fake.email(),
            "sexo": random.choice(["M", "F", "m", "f"]),
            "data_nascimento": nasc,
            "cidade": fake.city(),
            "estado": random.choice([fake.state_abbr(), fake.state(), fake.state_abbr().lower()]),
            "data_cadastro": fake.date_between("-3y"),
            "canal_aquisicao": random.choice(canais),
            "tipo_cliente": random.choice(tipos),
            "score_credito": random.randint(300, 900)
        })

    return clientes
```

```python
# 3. Gera o DataFrame
df_clientes = pd.DataFrame(gerar_clientes())
#df_clientes
```

```python
# 4. Salva o DataFrame como CSV no diretório de dados brutos
df_clientes.to_csv(
    RAWDIR / "clientes_raw.csv",
    index=False, encoding="utf-8")
```
