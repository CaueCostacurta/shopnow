```python
import sys
from pathlib import Path
import random
import pandas as pd
from faker import Faker
from datetime import datetime, timedelta

# Caminho real do notebook 
NOTEBOOK_DIR = Path.cwd() 

# Sobe até a raiz do projeto (shopnow/) 
ROOT = NOTEBOOK_DIR.parents[2] 

# Adiciona ao sys.path 
if str(ROOT) not in sys.path: sys.path.append(str(ROOT)) 

# Agora sim, pode importar 
from pipeline.utils import find_repo_root, get_raw_dir

# 1. Configuração do ambiente

ROOT = find_repo_root()
RAWDIR = get_raw_dir()
```

```python
# 2. Configuração do Faker, Seed e geração dos dados

# Inicializa o Faker e a seed para reprodutibilidade
fake = Faker("pt_BR")
random.seed(42)

def gerar_logistica(qtd: int = 5000) -> list[dict]:
    transportadoras = ["Correios", "correios", "Jadlog", "Total Express"]

    entregas = []

    for i in range(qtd):
        envio = fake.date_between("-2y")
        previsto = envio + timedelta(days=random.randint(2, 12))
        real = previsto + timedelta(days=random.randint(-4, 6))

        entregas.append({
            "id_entrega": i + 1,
            "id_pedido": f"PED-{random.randint(10000,99999)}",
            "data_envio": envio,
            "data_prevista": previsto,
            "data_entregue": real,
            "transportadora": random.choice(transportadoras),
            "custo_frete_real": round(random.uniform(8, 140), 2),
            "tentativas": random.randint(1, 4)
        })

    return entregas
```

```python
# 3. Gera o DataFrame
df_logistica = pd.DataFrame(gerar_logistica())
#df_logistica
```

```python
# 4. Salva o DataFrame como CSV no diretório de dados brutos
df_logistica.to_csv(
    RAWDIR / "logistica_raw.csv", 
    index=False, encoding="utf-8")
```
